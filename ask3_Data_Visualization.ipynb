```python
# Task 3: Interactive Dashboard with Customer Names
# CodeAlpha Internship - Data Analytics

import warnings
warnings.filterwarnings('ignore')

import numpy as np
import pandas as pd
import seaborn as sns
import plotly.express as px

# 1. Load Dataset
tips = sns.load_dataset('tips')
tips['tip_pct'] = ((tips['tip'] / tips['total_bill']) * 100).round(2)

# 2. Add Realistic Customer Names
first_names = ['Aarav', 'Ananya', 'Rahul', 'Priya', 'Rohan', 'Sneha', 'Amit', 'Neha',
               'Vikram', 'Pooja', 'Aditya', 'Divya', 'Karan', 'Riya', 'Arjun', 'Isha']
last_names = ['Sharma', 'Verma', 'Patel', 'Singh', 'Kumar', 'Gupta', 'Mehta', 'Joshi']

np.random.seed(42)  # Consistency ke liye
tips['customer_name'] = [
    f"{np.random.choice(first_names)} {np.random.choice(last_names)}"
    for _ in range(len(tips))
]

# 3. Interactive Scatter Plot (Customer Name included)
fig1 = px.scatter(
    tips,
    x='total_bill',
    y='tip',
    color='time',
    size='size',
    hover_name='customer_name',  # Title me Customer Name aayega
    hover_data={'customer_name': False, 'total_bill': ':.2f', 'tip': ':.2f', 'sex': True, 'day': True, 'size': True},
    title="🔍 Customer Spending & Tip Analysis (Hover to see Customer Name)",
    labels={'total_bill': 'Total Bill ($)', 'tip': 'Tip ($)'},
    template='plotly_white',
    color_discrete_sequence=['#2b5c8f', '#e76f51']
)
fig1.update_layout(height=480)
fig1.show()

# 4. Top 10 Highest Paying Customers (Bar Chart)
top_customers = tips.sort_values(by='total_bill', ascending=False).head(10)

fig2 = px.bar(
    top_customers,
    x='customer_name',
    y='total_bill',
    color='day',
    text_auto='.2f',
    hover_data=['tip', 'size', 'time'],
    title="🏆 Top 10 Highest Bill Paying Customers",
    labels={'customer_name': 'Customer Name', 'total_bill': 'Total Bill ($)'},
    template='plotly_white'
)
fig2.update_layout(height=480, xaxis_tickangle=-45)
fig2.show()
```
